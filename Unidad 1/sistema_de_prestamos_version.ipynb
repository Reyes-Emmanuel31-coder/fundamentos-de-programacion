```python
#SISTEMA DE PRESTAMOS
#Version 1 - Programacion basica y uso de listas
#Lista de alumnos activos en el sistemas
alumnos=["26040200","2604201", "26040202", "26040203","26040204"]
#Cada valor en la lista de computadoras tiene su lista correspondiente con el
#aula donde esta ubica, el sofware instalando y el estado de esa computadora
computadoras= ["PC-LC1-01",  "PC-LC1-02","PC-LC1-03",
               "PC-LC2-01","PC-LC2-02","PC-LC2-03",
               "PC-LC3-01","PC-LC3-02","PC-LC3-03","PC-LC3-04"]
aulas= ["LC1","LC1","LC1",
        "LC2","LC2","LC2",
        "LC3","LC3","LC3","LC3"]
sofware= ["Python, VS Code","Python,Java,VS Code", "AutoCAD, MALTAB",
          "Python, VS Code","Python,Java,VS Code","AutoCAD, MALTAB",
          "Python, VS Code","Python,Java,VS Code","AutoCAD, MALTAB","Python, VS Code"]
estados= ["Disponible","Prestada","Reservada",
         "Disponible","Prestada","Reservada",
         "Disponible","Prestada","Reservada","Mantenimiento"]
#MENU PRINCIPAL
#Usamos un ciclo while que se repita indefinidamente hasta que el usuario
#teclee la opcion 7 (salir) o cualquier numero superior a 7
opcion = 0
while opcion <= 6:
  print("\n================================")
  print("      SISTEMA DE PRESTAMOS")
  print("================================")
  print("1. Consultar computadoras")
  print("2. Buscar por sofware")
  print("3. Solicitar prestamo")
  print("4. Devolver computadora")
  print("5. Crear Reservacion")
  print("6. Agregar computadora")
  print("7. Salir")
  print("================================")
  opcion = int(input("Seleccione una opcion: "))
  #1. Consultar computadoras
  #si el usuario seleccione la opcion 1 se hace un ciclo (for) que recorre
  #toda la lista de computadoras e imprime los valores de las demas listas
  if opcion == 1:
    print("\n--- CONSULTAR COMPUTADORAS ---")
    for i in range(len(computadoras)):
      print("Aula:", aulas[i])
      print("ID:",computadoras[i])
      print("Sofware:", sofware[i])
      print("Estado:", estados[i])
      print("------------------------")
      # 2. Buscar por software
  elif opcion == 2:
      programa = input( "\nIngrese el software que desea buscar: ")
      print ("\n--- RESULTADOS - --")
      encontrado = False
      for i in range(len(computadoras)):
      # la funcion (lower) convierte a minuscula el texto, de esta forma
      # nos aseguramos que el texto a buscar y la informacion de las listas
      # este en minuscula y no tengamos problema de que no encunentra un programa
      # porque este escrito de forma diferente usando mayusculas y minusculas
       if programa.lower() in sofware[i].lower():
          print("ID:", computadoras[i])
          print ("Aula:", aulas[i])
          print("Software:", sofware[i])
          print("Estado:", estados[i])

          encontrado = True
      if encontrado == False:
          print("No se encontraron computadoras con ese software.")
# 3. Solicitar préstamo
  elif opcion == 3:
     print("\n--- SOLICITAR PRÉSTAMO ---")
     num_control = input( "\nIngrese el número de control: ")
     encontrado1 = False
     # Verificar si existe el alumno en la lista o BD
     if num_control in alumnos:
     # Si existe, continuamos con el procedimiento
       encontrado1 = True
       id_computadora = input("Ingrese el ID de la computadora: ")
       encontrado2 =False
     if not encontrado1:
        print("El alumno no existe.")
     else:
        id_computadora = input("Ingrese el ID de la computadora: " )
        encontrado2 = False
        computadora_index = -1
        for i in range(len(computadoras)):
            if computadoras [i] == id_computadora:
                encontrado2 = True
                computadora_index = i
                break # Found the computer, no need to continue loop

        if not encontrado2:
            print ("La computadora no existe.")
        elif estados [computadora_index] == "Disponible":
             estados [computadora_index] = "Prestada"
             print ("Préstamo registrado correctamente.")
             print ("Computadora:", computadoras[computadora_index])
             print ("Aula:", aulas[computadora_index])
        else:
            print("NO hay computadoras disponibles")
            else:
              print("el alumno no existe en la base de datos")
  elif opcion== 4:
    print("\n---DEVOLVER COMPUTADORA---")
    id_pc = input("Dame el ID de la computadora")
    encontrado = False
    for i in range(len(computadoras)):
      if computadoras[i] == id_pc:
       encontrado = True
      if estados [i]== "Prestada" or estados[i]=="Mantenimiento":
      estados [i]= "Disponible"
      print("La computadora", id_pc, "ahora esta disponible")
    else:
       print ("Esa computadora no esta prestada o reservada")
 if encontrado == False:
   print("No se encontro ese ID")
  # 6. Agregar computadora
  elif opcion == 6:
      print("\n--- AGREGAR COMPUTADORA -..")
      id_computadora = input( "ID de la computadora: ")
      aula = input( "Aula: ")
      programas = input ("Software instalado: ")
      computadoras. append (id_computadora)
      aulas.append(aula)
      sofware. append (programas)
      estados. append ("Disponible")
      print ("\nComputadora agregada correctamente.")
```
